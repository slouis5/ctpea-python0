## 0. Import et jeux de données

On utilise deux jeux de données classiques, intégrés à seaborn (chargés depuis le web) :
- **`tips`** : montant de l'addition et pourboire dans un restaurant (variables catégorielles + numériques)
- **`iris`** : mesures de fleurs (variables numériques uniquement)

```python
# imports
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
# jeux de données (chargés depuis le web)
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")

tips.head()
```

## Partie 1 — Matplotlib

**Matplotlib** est la librairie de visualisation historique de Python : plus verbeuse que seaborn, mais on retrouve ses graphiques dans (presque) toutes les autres librairies, y compris seaborn.

### 1. Courbe (`plt.plot`)

Le graphique de base pour visualiser une évolution (ex : une série temporelle) ou une fonction mathématique.

```python
# courbe simple
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y)
plt.title("Courbe (plt.plot)")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

### 2. Barres (`plt.bar`)

Pour comparer des valeurs entre plusieurs catégories.

```python
# addition moyenne par jour de la semaine
moyenne_par_jour = tips.groupby("day")["total_bill"].mean()

plt.figure(figsize=(8, 4))
plt.bar(moyenne_par_jour.index, moyenne_par_jour.values)
plt.title("Addition moyenne par jour (plt.bar)")
plt.xlabel("Jour")
plt.ylabel("Addition moyenne ($)")
plt.show()
```

### 3. Histogramme (`plt.hist`)

Pour visualiser la **distribution** d'une variable numérique.

```python
# distribution du montant des additions
plt.figure(figsize=(8, 4))
plt.hist(tips["total_bill"], bins=20)
plt.title("Distribution de l'addition (plt.hist)")
plt.xlabel("Montant de l'addition ($)")
plt.ylabel("Effectif")
plt.show()
```

### 4. Nuage de points (`plt.scatter`)

Pour visualiser la relation entre **deux variables numériques**.

```python
# relation entre l'addition et le pourboire
plt.figure(figsize=(8, 4))
plt.scatter(tips["total_bill"], tips["tip"])
plt.title("Addition vs pourboire (plt.scatter)")
plt.xlabel("Addition ($)")
plt.ylabel("Pourboire ($)")
plt.show()
```

### 5. Boîte à moustaches (`plt.boxplot`)

Pour résumer la distribution d'une variable (médiane, quartiles, valeurs extrêmes).

```python
# distribution de l'addition selon le moment du repas
groupes = [tips.loc[tips["time"] == t, "total_bill"] for t in tips["time"].unique()]

plt.figure(figsize=(8, 4))
plt.boxplot(groupes, tick_labels=tips["time"].unique())
plt.title("Addition par moment du repas (plt.boxplot)")
plt.ylabel("Addition ($)")
plt.show()
```

## Partie 2 — Seaborn

**Seaborn** s'appuie sur matplotlib mais propose une API plus simple, orientée DataFrame, avec de meilleurs réglages par défaut (couleurs, styles) et une gestion native des variables catégorielles via l'argument `hue`.

```python
# thème seaborn appliqué à tous les graphiques suivants
sns.set_theme(style="whitegrid")
```

### 6. Distribution (`sns.histplot`)

Équivalent de `plt.hist`, avec en plus la possibilité d'ajouter une estimation de densité (`kde=True`).

```python
# distribution de l'addition, par moment du repas
plt.figure(figsize=(8, 4))
sns.histplot(data=tips, x="total_bill", hue="time", kde=True)
plt.title("Distribution de l'addition (sns.histplot)")
plt.show()
```

### 7. Boîte à moustaches / violon (`sns.boxplot`, `sns.violinplot`)

Le violon combine la boîte à moustaches et la densité de la distribution.

```python
# boxplot : addition par jour
plt.figure(figsize=(8, 4))
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Addition par jour (sns.boxplot)")
plt.show()
```

```python
# violinplot : même idée, avec la forme de la distribution
plt.figure(figsize=(8, 4))
sns.violinplot(data=tips, x="day", y="total_bill")
plt.title("Addition par jour (sns.violinplot)")
plt.show()
```

### 8. Nuage de points avec catégorie (`sns.scatterplot`)

Comme `plt.scatter`, mais on peut colorer les points selon une variable catégorielle avec `hue`.

```python
# addition vs pourboire, coloré par moment du repas
plt.figure(figsize=(8, 4))
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time", style="smoker")
plt.title("Addition vs pourboire (sns.scatterplot)")
plt.show()
```

### 9. Barres agrégées (`sns.barplot`)

Contrairement à `plt.bar`, `sns.barplot` calcule directement la moyenne (et l'intervalle de confiance) à partir des données brutes.

```python
# addition moyenne par jour, avec intervalle de confiance
plt.figure(figsize=(8, 4))
sns.barplot(data=tips, x="day", y="total_bill", hue="time")
plt.title("Addition moyenne par jour et moment du repas (sns.barplot)")
plt.show()
```

### 10. Corrélations (`sns.heatmap`)

Pour visualiser rapidement les corrélations entre toutes les variables numériques d'un jeu de données.

```python
# matrice de corrélation du jeu de données iris
correlations = iris.drop(columns="species").corr()

plt.figure(figsize=(6, 5))
sns.heatmap(correlations, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Corrélations entre variables (sns.heatmap)")
plt.show()
```

### 11. Vue d'ensemble (`sns.pairplot`)

Affiche toutes les paires de variables numériques en un seul graphique : très utile pour une première exploration d'un jeu de données.

```python
# vue d'ensemble du jeu de données iris, coloré par espèce
sns.pairplot(iris, hue="species")
plt.show()
```
